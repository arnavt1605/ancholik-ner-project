```python
import matplotlib.pyplot as plt

# Data
models = ['BiLSTM-CRF', 'BiGRU-CRF', 'Char-CNN-BiLSTM-CRF', 'BanglaBERT']
within_f1 = [0.6601, 0.6855, 0.7089, 0.7770]
cross_f1 = [0.6058, 0.6415, 0.7414, 0.8348]

# Create figure and axis
fig, ax = plt.subplots(figsize=(8, 5))

# Draw the connecting lines first (so they stay behind the dots)
for i in range(len(models)):
    # Determine line color: light gray if performance dropped, dark gray if it improved
    line_color = '#d3d3d3' if cross_f1[i] < within_f1[i] else '#696969'
    ax.hlines(y=models[i], xmin=min(within_f1[i], cross_f1[i]), 
              xmax=max(within_f1[i], cross_f1[i]), color=line_color, linewidth=3, zorder=1)

# Plot the dots for Within-Dialect (Light Gray/White fill)
ax.scatter(within_f1, models, color='white', edgecolor='black', s=100, 
           label='Within-Dialect Mean F1', zorder=2, linewidth=1.5)

# Plot the dots for Cross-Dialect (Black fill)
ax.scatter(cross_f1, models, color='black', edgecolor='black', s=100, 
           label='Cross-Dialect Mean F1', zorder=2)

# Formatting and labels
ax.set_xlabel('F1 Score', fontsize=12)
ax.set_xlim(0.55, 0.85)

# Add a vertical grid to help read the exact F1 values
ax.xaxis.grid(True, linestyle='--', alpha=0.5)
ax.yaxis.grid(False) # Turn off y-axis grid to keep it clean
ax.set_axisbelow(True)

# Clean up spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['left'].set_visible(False)

# Position legend at the top
ax.legend(fontsize=10, loc='upper center', bbox_to_anchor=(0.5, 1.15), ncol=2, frameon=False)

# Save as vector graphic (PDF)
plt.tight_layout()
plt.savefig('dumbbell_f1_comparison.pdf', format='pdf', bbox_inches='tight')

# Display the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Data
dialects = ['Barishal', 'Chittagong', 'Mymensingh', 'Noakhali', 'Sylhet']
test_vocab = [3805, 4942, 3883, 4592, 4345]
covered_words = [2213, 1766, 2714, 2555, 2187]
coverage_pct = [58.16, 35.73, 69.89, 55.64, 50.33]

# Set up figure and axis
fig, ax = plt.subplots(figsize=(9, 6))
x = np.arange(len(dialects))
width = 0.5

# Plot 1: Total Test Vocabulary (Light Gray background bar)
ax.bar(x, test_vocab, width, label='Uncovered / Total Test Vocabulary', 
       color='#e0e0e0', edgecolor='black')

# Plot 2: Covered Words (Dark Gray foreground bar)
ax.bar(x, covered_words, width, label='Covered Words', 
       color='#525252', edgecolor='black')

# Add the Coverage Percentage as text labels above the bars
for i in range(len(dialects)):
    ax.text(x[i], test_vocab[i] + 100, f"{coverage_pct[i]}%", 
            ha='center', va='bottom', fontsize=11, fontweight='bold')

# Formatting and labels
ax.set_ylabel('Number of Words', fontsize=12)
ax.set_xticks(x)
ax.set_xticklabels(dialects, fontsize=11)

# Set Y-axis limit a bit higher to leave room for the percentage labels
ax.set_ylim(0, 5800)

# Position legend
ax.legend(fontsize=11, loc='upper left')

# Add horizontal gridlines for readability
ax.yaxis.grid(True, linestyle='--', alpha=0.7)
ax.set_axisbelow(True)

# Remove top and right spines
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)

# Save as vector graphic (PDF)
plt.tight_layout()
plt.savefig('vocabulary_coverage.pdf', format='pdf', bbox_inches='tight')

# Display the plot
plt.show()
```
